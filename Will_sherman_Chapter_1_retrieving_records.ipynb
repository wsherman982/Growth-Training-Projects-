```python
import pandas as pd

emp_df = pd.read_csv("emp.csv")
dept_df = pd.read_csv("dept.csv")

print(emp_df.head())
print(dept_df.head())
```

```python
from pandasql import sqldf

query = """
SELECT *
FROM emp_df
WHERE deptno = 10
   OR (comm IS NOT NULL AND sal <= 2000 AND deptno = 20)
"""

result = sqldf(query)
print(result)
```
